```python
# Supplied holdout assignment.
_, recipe_id = np.unique(X[:, :7], axis=0, return_inverse=True)
assert len(np.unique(recipe_id)) == 427
np.random.seed(SEED)
group_order = np.unique(recipe_id)
np.random.shuffle(group_order)
n_dev_groups = int(0.8 * len(group_order))
dev_groups = group_order[:n_dev_groups]
test_groups = group_order[n_dev_groups:]

dev_idx = []
test_idx = []
for i in range(len(recipe_id)):
    if recipe_id[i] in dev_groups:
        dev_idx.append(i)
    else:
        test_idx.append(i)
dev_idx = np.array(dev_idx, dtype=int)
test_idx = np.array(test_idx, dtype=int)
X_dev, y_dev, g_dev = X[dev_idx], y[dev_idx], recipe_id[dev_idx]
# Do not access test responses again until the final evaluation section.
```

**S1.** Does `0.8` apply to observations or recipe groups? Explain how `X[:, :7]` affects the split when a recipe is measured at several ages.

**Answer:** The `0.8` applies to **recipe groups**, not individual observations. `np.unique(X[:, :7], ...)` identifies unique ingredient combinations (ignoring age), assigning each observation a `recipe_id`. The code then shuffles the 427 unique recipe groups and assigns the first 80 % (`n_dev_groups = int(0.8 * 427) = 341`) to development. Every row belonging to the same ingredient recipe therefore stays in the same split—whether it was measured at 3 days or 28 days. This prevents data leakage: the model never trains on one age of a recipe and validates on another age of the same recipe.


```python
shuffled_groups = dev_groups.copy()
np.random.shuffle(shuffled_groups)
fold_groups = np.array_split(shuffled_groups, 5)

folds = []
fold_sizes = []
for groups in fold_groups:
    rows = []
    for i in range(len(g_dev)):
        if g_dev[i] in groups:
            rows.append(i)
    folds.append(np.array(rows, dtype=int))
    fold_sizes.append(len(rows))
fold_sizes = np.array(fold_sizes)
```

**S2.** Does `np.array_split(shuffled_groups, 5)` divide recipes or observations? Why can the five values in `fold_sizes` differ?

**Answer:** `np.array_split` divides **recipe groups** as evenly as possible into 5 parts. The five `fold_sizes` values count **observations** (rows) belonging to the recipes in each part. Because recipes differ in how many age measurements they have, equally-sized recipe partitions produce unequal observation counts. As confirmed by the output, fold sizes are [158, 167, 158, 172, 140], which differ because the 341 development recipes are not uniformly replicated across age values.


```python
splits = []
for k in range(5):
    val = folds[k]
    train_parts = []
    for j in range(5):
        if j != k:
            train_parts.append(folds[j])
    train = np.concatenate(train_parts)
    assert np.intersect1d(g_dev[train], g_dev[val]).size == 0
    splits.append((train, val))
assert np.intersect1d(dev_groups, test_groups).size == 0
assert np.array_equal(np.sort(np.concatenate(folds)), np.arange(len(X_dev)))
print(f"Development: {len(dev_idx)} rows / {len(dev_groups)} recipes")
print(f"Test reserved: {len(test_idx)} rows / {len(test_groups)} recipes")
print("Validation fold sizes:", fold_sizes)
print(f"Development strength range: {y_dev.min():.2f} to {y_dev.max():.2f} MPa")
```

## B. Preprocessing and metrics
**TODO 2:** return one mean and one population standard deviation per column.
If a column's standard deviation is 0, replace the corresponding value in `scale` with 1.

**TODO 3:** standardize `A` using the supplied `mean` and `scale`, without refitting.

**TODO 4:** return MSE as a scalar.

```python
def fit_standardizer(A):
    # TODO 2: return training means and safe scales (population std; replace 0 with 1)
    mean = A.mean(axis=0)
    scale = A.std(axis=0)
    scale[scale == 0] = 1.0
    return mean, scale

def apply_standardizer(A, mean, scale):
    # TODO 3: standardize using the supplied statistics without refitting
    return (A - mean) / scale

def mse(actual, predicted):
    # TODO 4
    return np.mean((actual - predicted) ** 2)

def rmse(actual, predicted):
    return np.sqrt(mse(actual, predicted))

# Small independent checks; neither example uses project test data.
toy = np.array([[2., 5.], [4., 5.], [6., 5.]])
toy_mean, toy_scale = fit_standardizer(toy)
assert np.allclose(toy_mean, [4., 5.]) and toy_scale[1] == 1
assert np.allclose(apply_standardizer(toy, toy_mean, toy_scale).mean(axis=0), 0)
assert np.allclose(toy_scale, [np.sqrt(8 / 3), 1.0])
check_mean, check_scale = fit_standardizer(np.array([[0.], [2.]]))
assert np.allclose(apply_standardizer(np.array([[3.], [5.]]), check_mean, check_scale), [[2.], [4.]])
assert np.isclose(mse(np.array([1., 4.]), np.array([2., 2.])), 2.5)
print("Preprocessing and metric checks passed.")
```

## C. Polynomial features and OLS
**TODO 5:** complete the fitting part of `fit_features`. Standardize raw inputs,
expand them with `PolynomialFeatures(degree=degree)`, and standardize the non-constant
expanded columns. Keep column 0 equal to 1. Define the arrays used by the supplied
state dictionary and return statement. The feature functions accept inputs of
shape `(n, p)`; the checks also use examples with fewer than eight input columns.

`apply_features` is supplied. The OLS, ridge, and lasso fitting functions return
coefficient vectors in matrix-column order, with the intercept first.

```python
def fit_features(X_train, degree):
    # TODO 5: fit raw-input standardizer, expand polynomial, standardize non-constant columns
    x_mean, x_scale = fit_standardizer(X_train)
    X_std = apply_standardizer(X_train, x_mean, x_scale)
    poly = PolynomialFeatures(degree=degree)
    P = poly.fit_transform(X_std)          # column 0 is always 1 (bias)
    feature_mean, feature_scale = fit_standardizer(P[:, 1:])  # fit on non-constant columns only
    P[:, 1:] = apply_standardizer(P[:, 1:], feature_mean, feature_scale)
    X_train_poly = P
    state = {"x_mean": x_mean, "x_scale": x_scale,
             "feature_mean": feature_mean, "feature_scale": feature_scale,
             "degree": int(degree)}
    return X_train_poly, state

# Supplied: reuse the saved degree and standardization parameters.
def apply_features(X_raw, state):
    U = apply_standardizer(X_raw, state["x_mean"], state["x_scale"])
    poly = PolynomialFeatures(degree=int(state["degree"]))
    # PolynomialFeatures.fit determines the feature layout only.
    P = poly.fit_transform(U)
    P[:, 1:] = apply_standardizer(P[:, 1:], state["feature_mean"], state["feature_scale"])
    return P

for degree, expected in zip(DEGREES, (9, 45, 165)):
    X_demo_poly, state_demo = fit_features(X_dev[:20], degree)
    assert X_demo_poly.shape[1] == expected
    assert np.allclose(X_demo_poly[:, 0], 1.0)
    assert np.allclose(X_demo_poly[:, 1:].mean(axis=0), 0.0, atol=1e-9)
    demo_scales = X_demo_poly[:, 1:].std(axis=0)
    assert np.all(np.isclose(demo_scales, 0.0) | np.isclose(demo_scales, 1.0))
    assert state_demo["degree"] == degree
    assert np.allclose(X_demo_poly, apply_features(X_dev[:20], state_demo), atol=1e-10)
    assert np.allclose(X_demo_poly[:1], apply_features(X_dev[:1], state_demo), atol=1e-10)
print("Polynomial column counts and saved transformations checked.")
```

**S4.** Suppose `apply_features` recalculated both sets of means and standard deviations from the validation or test inputs, instead of using the values saved in `state`. How would this affect predictions and evaluation?

**Answer:** Recalculating statistics from validation or test inputs would cause **data leakage from the target split into preprocessing**, producing artificially different feature scales for each split. During CV, validation predictions would be computed on features scaled differently from training features, making the coefficient vector learned on training data inapplicable—predictions and validation MSE would be wrong and incomparable across folds. On the test set the same problem applies: the model was fitted on training-derived scales, so applying test-derived scales breaks the mapping between raw inputs and the learned coefficients, yielding meaningless predictions and an over-optimistic or erratic test error estimate.


**S5.** What is the advantage of standardizing the non-constant polynomial features before ridge or lasso for degree 2 or 3, even though the raw inputs have already been standardized? Why leave column 0 unchanged?

**Answer:** Even after standardizing the raw inputs, polynomial expansion creates features (e.g., x₁², x₁x₂, x₁x₂x₃) whose variances differ dramatically—squared or cubed terms can have much larger scale than linear terms. Ridge and lasso add a penalty proportional to coefficient magnitude, so without re-standardizing the expanded features, the penalty shrinks coefficients of large-scale features more than small-scale ones, biasing regularization unfairly across feature families. Standardizing the non-constant polynomial features puts them on equal footing, so the penalty treats all features consistently. Column 0 (the bias/intercept column, always 1) is left unchanged because the intercept is excluded from the penalty by convention; standardizing a constant column would also produce division by zero.


**TODO 6:** return the OLS coefficient vector using a NumPy least-squares solver.
Do not form a matrix inverse.

```python
def fit_ols(X_train_poly, y_train):
    A = X_train_poly
    # TODO 6: use numpy least-squares solver (no explicit matrix inverse)
    beta, _, _, _ = np.linalg.lstsq(A, y_train, rcond=None)
    return beta

toy_poly = np.array([[1., -1.], [1., 0.], [1., 1.], [1., 2.]])
assert np.allclose(fit_ols(toy_poly, 2 + 3 * toy_poly[:, 1]), [2., 3.])
print("OLS check passed.")
```

## D. Ridge and lasso
Use the course objectives: $\mathrm{RSS}+\lambda\sum_{j=1}^p\beta_j^2$ for ridge and
$\mathrm{RSS}+\lambda\sum_{j=1}^p|\beta_j|$ for lasso. Exclude the intercept from both penalties.

**TODO 7:** return the ridge coefficient vector using augmented least squares and the supplied `D`.

**TODO 8:** fit `Lasso` and return its coefficient vector `beta`, with the intercept first.
Follow the Week 3 convention: omit the constant input column, use `fit_intercept=True`,
and set `alpha = lam / (2 * len(y_train))`. Use `LASSO_MAX_ITER` and `LASSO_TOL`.

```python
def fit_ridge(X_train_poly, y_train, lam):
    A = X_train_poly
    D = np.eye(A.shape[1])
    D[0, 0] = 0  # do not penalize the intercept
    # TODO 7: augmented least squares: stack [A; sqrt(lam)*D] and [y; 0]
    A_aug = np.vstack([A, np.sqrt(lam) * D])
    y_aug = np.concatenate([y_train, np.zeros(A.shape[1])])
    beta, _, _, _ = np.linalg.lstsq(A_aug, y_aug, rcond=None)
    return beta

def fit_lasso(X_train_poly, y_train, lam):
    # TODO 8: sklearn Lasso with course convention
    # Objective: RSS + lam*sum|beta_j|  (j>=1, intercept excluded)
    # sklearn Lasso minimises RSS/(2n) + alpha*sum|beta_j|  => alpha = lam/(2n)
    n = len(y_train)
    alpha = lam / (2 * n)
    clf = Lasso(alpha=alpha, fit_intercept=True,
                max_iter=LASSO_MAX_ITER, tol=LASSO_TOL)
    clf.fit(X_train_poly[:, 1:], y_train)  # omit constant column; intercept fitted separately
    beta = np.concatenate([[clf.intercept_], clf.coef_])
    return beta

# Supplied: a common interface keeps the CV loop unchanged across methods.
def fit_coefficients(X_train_poly, y_train, method, lam):
    if method == "ols":
        return fit_ols(X_train_poly, y_train)
    if method == "ridge":
        return fit_ridge(X_train_poly, y_train, lam)
    if method == "lasso":
        with warnings.catch_warnings(record=True) as recorded:
            warnings.simplefilter("always")
            fitted = fit_lasso(X_train_poly, y_train, lam)
        for warning in recorded:
            if warning.category.__name__ == "ConvergenceWarning":
                raise RuntimeError(
                    f"Lasso did not converge (lambda={lam}, features={X_train_poly.shape[1] - 1}). "
                    "Keep the test set unopened and report the warning to your instructor. "
                    + str(warning.message))
            warnings.warn(warning.message, warning.category)
        return fitted
    raise ValueError("Unknown model family")

# Checks
assert np.allclose(fit_ridge(toy_poly, np.full(4, 7.), 100.), [7., 0.], atol=1e-8)
toy_beta = fit_lasso(toy_poly, 2 + 3 * toy_poly[:, 1], 0.1)
assert np.isfinite(toy_beta).all() and toy_beta.shape == (2,)
penalty_X = np.array([[1., -1.], [1., 1.]])
penalty_y = np.array([-2., 2.])
assert np.allclose(fit_ridge(penalty_X, penalty_y, 2.), [0., 1.], atol=1e-8)
assert np.allclose(fit_lasso(penalty_X, penalty_y, 4.), [0., 1.], atol=1e-5)
assert np.allclose(fit_lasso(penalty_X, np.array([6., 8.]), 4.), [7., 0.], atol=1e-5)
print("Regularization checks passed.")
```

## E. Cross-validation and model selection
**TODO 9:** complete `score_fold`. Fit preprocessing and coefficients using only
`X_train` and `y_train`, then define `pred_train` and `pred_val`.
Use `fit_features`, `apply_features`, and `fit_coefficients`.
The function returns training and validation MSE.

```python
def score_fold(X_train, y_train, X_val, y_val, degree, method, lam):
    # TODO 9: fit preprocessing and coefficients on training data only, predict on both
    X_train_poly, state = fit_features(X_train, degree)
    beta = fit_coefficients(X_train_poly, y_train, method, lam)
    pred_train = X_train_poly @ beta
    X_val_poly = apply_features(X_val, state)
    pred_val = X_val_poly @ beta
    return mse(y_train, pred_train), mse(y_val, pred_val)

# Small check with a deliberately imperfect validation response.
check_X_train = np.array([[0.], [1.], [2.]])
check_y_train = np.array([1., 3., 5.])
check_X_val = np.array([[3.]])
check_y_val = np.array([9.])
check_train_mse, check_val_mse = score_fold(
    check_X_train, check_y_train, check_X_val, check_y_val, 1, "ols", 0.0)
assert np.isclose(check_train_mse, 0.0, atol=1e-9)
assert np.isclose(check_val_mse, 4.0, atol=1e-9)
print("Single-fold check passed.")
```

```python
# Supplied candidate order also determines how exact ties are resolved.
candidates = []
for degree in DEGREES:
    candidates.append((degree, "ols", 0.0))
    for lam in RIDGE_LAMBDAS:
        candidates.append((degree, "ridge", lam))
    for lam in LASSO_LAMBDAS:
        candidates.append((degree, "lasso", lam))
assert len(candidates) == 21

# Supplied baseline.
baseline_fold_mse = []
for train, val in splits:
    baseline_predictions = np.full(len(val), y_dev[train].mean())
    baseline_fold_mse.append(mse(y_dev[val], baseline_predictions))
baseline_fold_mse = np.array(baseline_fold_mse)
```

**Baseline prediction.** In one CV fold, what does this baseline predict for each validation observation? Which data determine that prediction, and does it depend on the observation's input features?

**Answer:** In each CV fold, the baseline predicts the **mean of the training-fold response values** (`y_dev[train].mean()`) as a single constant for every validation observation. The prediction depends only on the target values of the training fold observations, not on any input features of the validation observations. Every validation observation receives the same predicted value regardless of its ingredient recipe or age.


**TODO 10:** implement `run_cv(X, y, splits, candidates)` using `score_fold`.
Return two NumPy arrays, `train_mse_cv` and `val_mse_cv`, each with shape
`(len(candidates), len(splits))`. Row `i` corresponds to candidate `i`; column `k`
corresponds to fold `k`. Use the supplied indices with this function's `X` and `y`.

```python
def run_cv(X, y, splits, candidates):
    # TODO 10: evaluate every candidate on every fold
    n_cand = len(candidates)
    n_folds = len(splits)
    train_mse_cv = np.zeros((n_cand, n_folds))
    val_mse_cv   = np.zeros((n_cand, n_folds))
    for i, (degree, method, lam) in enumerate(candidates):
        for k, (train, val) in enumerate(splits):
            t_mse, v_mse = score_fold(
                X[train], y[train], X[val], y[val], degree, method, lam)
            train_mse_cv[i, k] = t_mse
            val_mse_cv[i, k]   = v_mse
    return train_mse_cv, val_mse_cv

# Two candidates and two folds with independently known errors.
check_X = np.array([[0.], [1.], [2.], [3.]])
check_y = np.array([1., 3., 5., 9.])
check_splits = [
    (np.array([0, 1]), np.array([2, 3])),
    (np.array([2, 3]), np.array([0, 1]))
]
check_candidates = [(1, "ols", 0.0), (1, "ridge", 2.0)]
check_train, check_val = run_cv(check_X, check_y, check_splits, check_candidates)
assert check_train.shape == (2, 2) and check_val.shape == (2, 2)
assert np.allclose(check_train, [[0., 0.], [0.25, 1.]], atol=1e-8)
assert np.allclose(check_val, [[2., 10.], [11.25, 1.]], atol=1e-8)
print("CV workflow check passed.")

train_mse_cv, val_mse_cv = run_cv(X_dev, y_dev, splits, candidates)
assert train_mse_cv.shape == (21, 5) and val_mse_cv.shape == (21, 5)
assert np.isfinite(train_mse_cv).all() and np.isfinite(val_mse_cv).all()
assert (train_mse_cv >= 0).all() and (val_mse_cv >= 0).all()
```

**TODO 11:** implement `select_model(val_mse_cv)`. Return the equally weighted
mean validation MSE for every candidate as a 1D array, followed by the index of
the candidate with the lowest mean. Choose the first candidate in an exact tie.
Use only validation scores for selection. Report CV RMSE as the square root of
mean fold MSE; keep the test set reserved.

```python
def select_model(val_mse_cv):
    # TODO 11: equally weighted mean of fold MSEs; first minimum wins ties
    mean_mse = val_mse_cv.mean(axis=1)
    best_index = int(np.argmin(mean_mse))
    return mean_mse, best_index

check_scores = np.array([[1., 9., 8.], [5., 4., 6.], [9., 1., 8.]])
check_means, check_best = select_model(check_scores)
assert np.allclose(check_means, [6., 5., 6.]) and check_best == 1
_, check_tie = select_model(np.array([[3., 3.], [3., 3.]]))
assert check_tie == 0

mean_mse, best_index = select_model(val_mse_cv)
best_degree, best_method, best_lam = candidates[best_index]
linear_index = candidates.index((1, "ols", 0.0))

# Supplied result formatting.
print(f"Mean-response baseline: CV RMSE = {np.sqrt(baseline_fold_mse.mean()):.3f} MPa")
print(f"{'family':<8} {'degree':>6} {'lambda':>9} {'train RMSE*':>12} {'CV RMSE':>10} {'MSE range':>11}")
for i in range(len(candidates)):
    degree, method, lam = candidates[i]
    print(f"{method:<8} {degree:>6} {lam:>9g} "
          f"{np.sqrt(train_mse_cv[i].mean()):>12.3f} {np.sqrt(mean_mse[i]):>10.3f} "
          f"{np.ptp(val_mse_cv[i]):>11.3f}")
print("* train RMSE: square root of mean training-fold MSE; range units: MPa^2")
print("Selected:", best_method, "degree =", best_degree, "lambda =", best_lam)
print("Selected fold MSE:", val_mse_cv[best_index])
```

**R1 — Selection.** Select a model using your CV results and justify your choice. Record your choice before testing.

**Answer:** I select **ridge regression with degree 2 and λ = 0.1** (CV RMSE = 8.020 MPa). This candidate achieves the lowest mean validation MSE across all 21 candidates. Degree-3 OLS dramatically overfits (training RMSE = 4.0 MPa but CV RMSE = 12.8 MPa), confirming that more regularization is needed at that degree. ## F. Final refit, prediction, and test
**TODO 13:** implement `fit_final`. Fit preprocessing and coefficients on all supplied
`X_fit` and `y_fit`, using `fit_features` and `fit_coefficients`.
Return the feature state with the coefficient vector stored as `state["beta"]`.

**TODO 14:** complete `predict(raw8, model_state)`. Return a strength vector of shape
`(n,)` for raw inputs of shape `(n, 8)`, using the saved degree, scaling statistics,
and coefficients without re-estimating statistics or coefficients.

```python
def fit_final(X_fit, y_fit, degree, method, lam):
    # TODO 13: fit preprocessing and coefficients on all supplied data; store beta in state
    X_poly, state = fit_features(X_fit, degree)
    beta = fit_coefficients(X_poly, y_fit, method, lam)
    state["beta"] = beta
    return state

def predict(raw8, model_state):
    raw8 = np.asarray(raw8, dtype=float)
    if raw8.ndim != 2 or raw8.shape[1] != 8 or not np.isfinite(raw8).all():
        raise ValueError("Expected a finite numeric array of shape (n, 8).")
    # TODO 14: apply saved feature transformation and dot with saved coefficients
    X_poly = apply_features(raw8, model_state)
    return X_poly @ model_state["beta"]

# Independent complete-workflow check; these are not project test observations.
check_final_X = np.zeros((4, 8))
check_final_X[:, 0] = [0., 1., 2., 4.]
check_final_y = np.array([1., 3., 5., 9.])
check_model = fit_final(check_final_X, check_final_y, 1, "ols", 0.0)
assert np.allclose(check_model["x_mean"], check_final_X.mean(axis=0))
assert np.allclose(predict(check_final_X, check_model), check_final_y, atol=1e-8)
check_new = np.zeros((2, 8))
check_new[:, 0] = [3., 5.]
assert np.allclose(predict(check_new, check_model), [7., 11.], atol=1e-8)
assert np.allclose(predict(check_new[:1], check_model), predict(check_new, check_model)[:1])
print("Refit and reusable prediction checks passed.")
```

Use `predict` for both models. This is the fixed final comparison; do not revise

```python
X_test, y_test = X[test_idx], y[test_idx]
# TODO 15: refit selected model and degree-1 OLS reference on all development data
final_state     = fit_final(X_dev, y_dev, best_degree, best_method, best_lam)
reference_state = fit_final(X_dev, y_dev, 1, "ols", 0.0)
pred_test       = predict(X_test, final_state)
pred_reference  = predict(X_test, reference_state)
test_mse        = mse(y_test, pred_test)
reference_test_mse = mse(y_test, pred_reference)

# Supplied checks and reporting.
assert np.allclose(final_state["x_mean"], X_dev.mean(axis=0))
assert np.allclose(reference_state["x_mean"], X_dev.mean(axis=0))
assert int(final_state["degree"]) == best_degree and int(reference_state["degree"]) == 1
assert pred_test.shape == y_test.shape and pred_reference.shape == y_test.shape
assert np.isfinite(pred_test).all() and np.isfinite(pred_reference).all()
assert np.isfinite(test_mse) and np.isfinite(reference_test_mse)
assert test_mse >= 0 and reference_test_mse >= 0
print(f"Selected model: test MSE = {test_mse:.3f} MPa^2; RMSE = {np.sqrt(test_mse):.3f} MPa")
print(f"Linear reference: test MSE = {reference_test_mse:.3f} MPa^2; RMSE = {np.sqrt(reference_test_mse):.3f} MPa")
print("Test observations:", len(y_test))
```

**R3 — Test error.** What do the test error and residual plot tell you about the model?

**Answer:** The test RMSE of 8.748 MPa is slightly worse than the CV RMSE of 8.020 MPa, consistent with normal generalization variance rather than overfitting. The residual plot, however, reveals two features: (1) a slight fan-shape—residuals are larger in absolute value at higher predicted strengths—suggesting mild heteroscedasticity; and (2) a small negative mean residual (≈ –1.2 MPa), indicating a very slight tendency to over-predict.
